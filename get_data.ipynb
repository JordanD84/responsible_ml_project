# FLORES+

The dataset can be [accessed through on HuggingFace](https://huggingface.co/datasets/openlanguagedata/flores_plus). 

This is **Gated dataset**, meaning you have to log in and accept some terms and conditions. Additionally, in the IDE, you have to login to your hugging face account via env. variables or `from huggingface_hub import login`.

Since the dataset's intended usage is for MT evaluation and not training, the developpers ask that the dataset may not be made publicly available. See HuggingFace website for more information on [Gated dataset](https://huggingface.co/docs/hub/en/datasets-gated).

```python
# Load dataset from HuggingFace
ds = load_dataset("openlanguagedata/flores_plus", split="devtest")
# Convert to pandas
df = pd.DataFrame(ds)
# Cleaning
flores= (df
 .loc[df.iso_639_3.isin(['eng','fra']), ['id','iso_639_3', 'text', 'domain', 'topic']] # only keeping some columns
 .reset_index(drop=True)
 .pivot(index=['id', 'domain', 'topic'], columns='iso_639_3', values=['text']) # long to wide
 )
flores.columns = ['_'.join(col) if isinstance(col, tuple) else col for col in flores.columns] 
flores.reset_index(inplace=True) # removing multi-index
```

```python
# export
flores.to_parquet('data/test_set/flores_plus.parquet', index=False)
```
